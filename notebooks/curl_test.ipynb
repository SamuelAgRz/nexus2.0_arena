```python
import pandas as pd
df_golden_set= pd.read_csv('GoldenSetNSR(Sheet1).csv')
df_golden_set
```

```python
import requests
import json
from tqdm import tqdm

url = "https://emt-ai-pbi-agent.azurewebsites.net/et/api/pbi-data-xmla-service"

headers = {
    "client_Id": "",
    "client_secret": "",
    "tenant_Id": "",
    "workspace": "NSR LATAM",
    "catalog": "NSR LATAM Cube",
    "aut_url": "https://login.microsoftonline.com/",
    "scope": "https://analysis.windows.net/powerbi/api/.default",
    "user_id": "rcasianomorales@coca-cola.com",
    "Content-Type": "application/json",
    "Authorization": "",
}
```

```python
#Create a dataframe to store the results, including response time and status code
results_df = []
for i in tqdm(range(len(df_golden_set))):
	query = df_golden_set['Query'][i]
	payload = {
		"query": query,
		"msg": {}
	}
	response = requests.post(url, headers=headers, json=payload)
	results_df.append({
		"Query": query,
		"Status Code": response.status_code,
		"Response Time": response.elapsed.total_seconds(),
		"Response": response.json()
	})
results_df= pd.DataFrame(results_df)
results_df
```

```python
results_df.to_csv('golden_set_results.csv', index=False)
```
